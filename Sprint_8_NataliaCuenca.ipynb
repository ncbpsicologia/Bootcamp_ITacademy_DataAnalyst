```python
# Revisión de la columna Cognoms
df["Cognoms"].unique()

# Se observa un patron en los errores relacionados con las tildes, del estilo: Ã± → ñ, Ã³ → ó, Ã¡ → á, Ã© → é, Ã¼ → ü
# En el paso anterior, en la columna "Nom" se había detectado una fila inventada "???"" y una vacía "nan", que se mantiene
```

```python
# Corrección de errores de codificación en la columna Cognoms

def corregir_codificacion(texto):
    if isinstance(texto, str):
        try:
            return texto.encode("latin1").decode("utf-8")
        except:
            return texto
    return texto

df["Cognoms"] = df["Cognoms"].apply(corregir_codificacion)

df["Cognoms"].unique()
```

```python
# » Taula resum que permeti veure el sou mig, medià, mínim i màxim per Gènere. Ordena la taula en funció del sou mig.

# Agrupo por Género mediante groupby, nos fijamos en el Salario (concretamente, media, mediana, mínimo y máximo)
# Para varias operaciones a la vez, usamos agg()
# Para ordenar la tabla en función del salario medio, usamos sort_values. Ascending=False ya que por defecto es True (ascendente)

tabla_genere = (
    df.groupby("Gènere")["Salari_mensual"]
      .agg(
          Salari_Mig="mean",
          Mediana="median",
          Mínim="min",
          Màxim="max"
      )
      .sort_values(by="Salari_Mig", ascending=False)
)

tabla_genere
```

```python
# » Crear una taula resum amb el salari mig per gènere (files) i país d'origen (columnes).
# Afegeix-hi les mitjanes als marges de la taula.

# Para la tabla cruzada empleamos pivot_table
# Para las medianas en los márgenes, margins=True (como crea la columna "All", decido ponerle un nombr emás intuitivo)")

tabla_pais = pd.pivot_table(
    df,
    values="Salari_mensual",
    index="Gènere",
    columns="País_d'origen",
    aggfunc="mean",
    margins=True,
    margins_name="Mitjana"
)

tabla_pais
```

## Exercici 2: Optimització de rutes (problema aplicat)
🌍 Context: L'empresa no disposa d'un control fiable del quilometratge dels desplaçaments dels seus treballadors. Abans d'implementar un sistema avançat, es vol fer una prova pilot amb treballadors ubicats a Espanya.
Per a aquesta prova:
» Es treballa amb una matriu de distàncies entre ciutats.
»Es vol una solució simple, raonable i explicable.
» No es busca la ruta òptima, sinó una aproximació coherent.
Carrega l'arxiu matriu_distancies.xlsx a pandas de manera que les files i columnes siguin les ciutats i s'eliminin "Las Palmas de Gran Canaria" i "Palma".
Desenvolupa una funció que:
» Rebi: una ciutat d'origen i una ciutat de destinació.
» Retorni: l'ordre de les ciutats visitades (ruta) i la distància total recorreguda.
La ruta s'ha de construir seguint una estratègia senzilla: anar sempre a la ciutat no visitada més propera fins a arribar a la destinació.


```python
# importamos el nuevo archivo

df_distancias = pd.read_excel("matriu_distancies.xlsx", index_col=0) #el último paso es para que la 1a columna sea el índice

df_distancias
```

```python
#ahora eliminamos las ciudades indicadas

df_distancias = df_distancias.drop(
    ["Las Palmas de Gran Canaria", "Palma"],
    axis=0
)

df_distancias = df_distancias.drop(
    ["Las Palmas de Gran Canaria", "Palma"],
    axis=1
)

#comprobamos

df_distancias
```

```python
# defino la funcion

def ruta(origen, destino):

    ciudad_actual = origen  # Guardamos la ciudad en la que estamos actualmente
    ciudades_visitadas = [origen]  # Creo una lista con las ciudades que ya hemos visitado

    distancia_total = 0  # empiezo un contador de kms des de 0

    while ciudad_actual != destino:  # Repetimos mientras la ciudad actual no sea el destino

        distancias = df_distancias.loc[ciudad_actual]  # Cogemos las distancias desde la ciudad actual
        no_visitadas = distancias.drop(ciudades_visitadas)  # eliminamos las ciudades que ya hemos visitado
        siguiente_ciudad = no_visitadas.idxmin()  # De las restantes, elige la ciudad no visitada más cercanna
        distancia = no_visitadas.min()  # Obtenemos la distancia hasta esa ciudad
        ciudades_visitadas.append(siguiente_ciudad)  # Añadimos la ciudad a la ruta
        distancia_total = distancia_total + distancia  # Sumamos la distancia al total
        ciudad_actual = siguiente_ciudad  # Nos desplazamos a la nueva ciudad

    return ciudades_visitadas, distancia_total  # Devolvemos la ruta y la distancia total
```
