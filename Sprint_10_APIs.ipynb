```python
''' Queremos:
Llista de publicacions (/posts)
Llista d’usuaris (/users)
Llista de tasques (/todos)
'''

print("LLISTA DE PUBLICACIONS")
url = "https://jsonplaceholder.typicode.com/posts"
response_posts = requests.get(url)
posts = response_posts.json()
print(posts)

print()

print("LLISTA D'USUARIS")
url = "https://jsonplaceholder.typicode.com/users"
response_users = requests.get(url)
users = response_users.json()
print(users)

print()

print("LLISTA DE TASQUES")
url = "https://jsonplaceholder.typicode.com/todos"
response_todos = requests.get(url)
todos = response_todos.json()
print(todos)
```

4. Fes una petició POST per crear una nova publicació fictícia.

Inclou el títol, el cos del missatge i un userId.

- Mostra la resposta JSON.
- Mostra el codi d'estat.


```python
import requests

print("CREAR NUEVA PUBLICACIÓN")

# indicar donde queremos crear la nueva publicación
url = "https://jsonplaceholder.typicode.com/posts"

# creamos los datos de la publicación
nou_post = {
    "title": "Natàlia practicando",
    "body": "El cuerpo de la publicación fictícia donde explicaría lo que estoy haciendo para aprender.",
    "userId": 1
}

# realizamos el post (enviar datos, a diferenica de antes con el get que los pedíamos)
response_post = requests.post(url, json=nou_post)

# ver la respuesta de la API
print("Resposta JSON:")
print(response_post.json())

print()

# ver el código de estado
print("Codi d'estat:")
print(response_post.status_code)
```


5. Fes una petició PATCH per modificar parcialment una publicació existent.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
import requests

print("MODIFICAR MI PUBLICACIÓN")

# indicar donde está la publicación que queremos modificar (añadimos el /1 porque es el id que queremos modificar)
url = "https://jsonplaceholder.typicode.com/posts/1"

# modificamos los datos de la publicación
cambio = {
    "title": "Natàlia aprendiendo",
}

# realizamos el patch ()
response_patch = requests.patch(url, json=cambio)

# ver la respuesta de la API
print("Resposta JSON:")
print(response_patch.json())

print()

# ver el código de estado
print("Codi d'estat:")
print(response_patch.status_code)
```

6. Fes una petició DELETE sobre una publicació.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
import requests

print("ELIMINAR PUBLICACIÓn")

# indicamos donde está la publicación que queremos borrar
url = "https://jsonplaceholder.typicode.com/posts/1"

#ejecutamos el delete
response_delete = requests.delete(url)

# ver la respuesta d ela API
print("Resposta JSON:")
print(response_delete.json())

print()

# ver el código de estado
print("Codi d'estat:")
print(response_delete.status_code)
```

Documentación de la API MusicBrainz

**ENDPOINTS**

Dos de los endpoints disponibles son:

- `/ws/2/artist` → buscar info sobre artistas (como he hecho en el ejemplo)
- `/ws/2/recording` → buscar info sobre canciones



**PARÁMETROS Y FILTROS**

Esta API pemite utilizar estos parámetros en las búsquedas:

- `query` → indica qu queremos buscar
- `limit` → indica el nº max. de resultados que queremos obtener (entre 1 y 100)
- `offset` → permite empezar los resultados desde una posición
- `fmt` → permite escoger el formato de respuesta (indicando por ejemplo `json`como he hehco en el ejemplo de JB)

En les cerques d'artistes també podem utilitzar camps com `artist`, `country`, `type`, `gender`, `area` o `tag` per filtrar els resultats.



**FORMATO DE RESPUESTA**

Podemos obtener el formato json usando el paràmetre `fmt=json`.
Se comprueba usando, en este caso, `response_music.json()`

```python
# Voy a pedir información sobre Justin Bieber

print("INFORMACIÓN DE JUSTIN BIEBER")

url = "https://musicbrainz.org/ws/2/artist/?query=artist:Justin%20Bieber&fmt=json"

response_music = requests.get(url)
music = response_music.json()

# Código de estado
print("Codi d'estat:", response_music.status_code)

# Algunos campos de la respuesta JSON
print("Nombre de resultados:", music["count"])
print("Primer artista encontrado:", music["artists"][0]["name"])
```
