```python
# PASO 1: OBTENER DATOS

# El objetivo de esta función es obtener datos básicos sobre la persona usuaria.
# La función recibirá los datos a través de inputs (un input por dato), los almacenará en variables y retornará los datos introducidos en un diccionario.
# Concretamente, se pide: nombre y apellidos, teléfono de contacto, años de experiencia en la empresa, nivel de catalán (entre 0 y 3), nivel de Excel (entre 0 y 3) y nivel de titulación académica (entre 0 y 3)

def obtenir_dades():
    nom = input("Escriu el teu nom i cognoms: ")
    tlf = input("Telèfon de contacte: ")
    exp = int(input("Introdueix quants anys d'experiència tens dins l'empresa: "))
    cat = int(input(
        "Indica el teu nivell de català (0-3)\n"
        "0 = sense coneixement\n"
        "1 = fins a nivell B1\n"
        "2 = nivell B2\n"
        "3 = nivell C1 o superior\n"
        "Valor: "
        ))
    exl = int(input(
        "Indica el teu nivell de domini d'Excel (0-3)\n"
        "0 = nivell bàsic\n"
        "1 = ús de funcions\n"
        "2 = taules dinàmiques\n"
        "3 = Visual Basic i Macros\n"
        "Valor: "
        ))
    tit = int(input(
        "Indica el teu nivell acadèmic (0-3)\n"
        "0 = ESO\n"
        "1 = FP\n"
        "2 = Universitat\n"
        "3 = Màster o superior\n"
        "Valor: "
       ))

    candidatura = {
        "nom": nom,
        "telefon": tlf,
        "experiencia": exp,
        "catala": cat,
        "excel": exl,
        "titulacio": tit
        }

    return candidatura
```

```python
# COMPROBACIÓN DEL PASO 1

obtenir_dades()
```

```python
# PASO 2: VALIDAR DATOS

# El objetivo de esta función es validar los datos obtenidos, comprobando que sean correctos/adecuados. Recibe el diccionario que hemos creado.
# La función comprobará distintas condiciones a través de condicionales. Si algun dato es incorrecto, rechazará el registro. Si todos los datos son correctos, retornará que la validación es correcta.
# Concretamente, se comprobará que los años de experiencia no sean negativos, y que el nivel de catalan, de Excel y de titulación académica tengan un valor entre 0 y 3

def validar_dades(candidatura):
    if candidatura["experiencia"] < 0:
        print("Error: els anys d'experiència no poden ser negatius.")
        return False

    if candidatura["catala"] < 0 or candidatura["catala"] > 3:
        print("Error: el nivell de català ha de ser un valor entre 0 i 3.")
        return False

    if candidatura["excel"] < 0 or candidatura["excel"] > 3:
        print("Error: el nivell d'Excel ha de ser un valor entre 0 i 3.")
        return False

    if candidatura["titulacio"] < 0 or candidatura["titulacio"] > 3:
        print("Error: el nivell de titulació ha de ser un valor entre 0 i 3.")
        return False

    print("\nRegistre acceptat.")
    return True
```

```python
# PASO 3: CALCULAR PUNTUACIÓN

# El objetivo de esta función es calcular la puntuación de la persona usuaria. Recibe el diccionario que hemos creado.
# La función realizará los cálculos pertinentes a través de los operadores y retornará la puntuación total
# Concretamente, la puntuación se basará en los años de experiencia en la empresa (con un peso significativo), el nivel de catalán, de dominio de Excel y de titulación académica.

def calcular_puntuacio(candidatura):
  puntuacio = (
      candidatura["experiencia"] * 1.5
      + candidatura["catala"]
      + candidatura["excel"]
      + candidatura["titulacio"]
  )

  return puntuacio
```

```python
# COMPROBACIÓN DEL PASO 3

puntuacio = calcular_puntuacio(candidatura)
print("La puntuació és:", puntuacio)
candidatura["puntuacio"] = puntuacio
```

```python
# PASO 4: CLASIFICAR CANDIDATOS

# El objetivo de esta función es asignar una etiqueta a cada persona usuaria en base a su puntuacion, clasificarlas. Recibe la puntuacion calculada en el paso anterior.
# La función asignará y retornará la etiqueta correspondiente a través de condicionales.
# Concretamente, las etiquetas seran: Prioritat Alta → puntuació ≥ 12, Candidat adequat → 8 ≤ puntuació < 12, No prioritari → puntuació < 8

def clasificar_candidat(puntuacio):
  if puntuacio >= 12:
    return "Prioritat alta"

  elif puntuacio >= 8:
    return "Candidat/da adequat/da"

  else:
    return "No prioritat"
```

```python
# COMPROBACIÓN 1 DEL PASO 4

classificacio = clasificar_candidat(puntuacio)
print("L'etiqueta de classificació és:", classificacio)
candidatura["classificacio"] = classificacio
```

```python
# PASO 5: MOSTRAR INFORMACIÓN INDIVIDUAL

# El objetivo de esta función es mostrar un resumen del resultado de la candidatura. Recibe el diccionario que hemos creado (para la variable nom) y las variables puntuacio y clasificacio creadas después.
# La función mostrará en pantalla concretamente los siguientes datos: nombre del/a candidato/a, puntuación obtenida y etiqueta de clasificación asignada.

def mostrar_individual(candidatura, puntuacio, classificacio):
    print("--RESULTAT DEL/A CANDIDAT/A--")
    print(f"Nom: {candidatura['nom']}")
    print(f"Puntuació: {puntuacio}")
    print(f"Classificació: {classificacio}")
```

```python
# COMPROBACIÓN DEL PASO 5

mostrar_individual(candidatura, puntuacio, classificacio)
```

```python
# PASO 6: MOSTRAR INFORME

# El objetivo de esta función es mostrar los candidatos por orden de prioridad. Recibe el listado de candidaturas (un diccionario por cada persona)
# La función recorrerá los registros con un bucle for, ordenará los candidatos y los mostrará en pantalla, concretamente los siguientes datos: nombre del/a candidato/a, teléfono, puntuación obtenida y etiqueta de clasificación asignada.

def mostrar_informe(listado_candidaturas):

    listado_candidaturas.sort(key=lambda c: c["puntuacio"], reverse=True)

    print("\n===== INFORME FINAL =====\n")

    for c in listado_candidaturas:
        print(f"Nom: {c['nom']}")
        print(f"Telèfon: {c['telefon']}")
        print(f"Puntuació: {c['puntuacio']}")
        print(f"Classificació: {c['classificacio']}")
        print("-------------------")
```

```python
# PASO 7: FUNCIÓN MAIN (ORQUESTADORA)

# El objetivo de esta función es orquestar el resto de pasos. Como el código es largo, los voy a ir describiendo a lo largo de la función, por facilitar la legibilidad:

def main_seleccio():

    # Crear una lista vacía para guardar las candiadturas válidas
    llista_candidats = []

    # Repetir el proceso hasta obtener 10 candidaturas válidas
    while len(llista_candidats) < 10:

        # PASO 1: obtener datos
        candidatura = obtenir_dades()

        # PASO 2: validar datos
        if validar_dades(candidatura):

            # PASO 3: calcular puntuación
            puntuacio = calcular_puntuacio(candidatura)

            # PASO 4: clasificar
            classificacio = clasificar_candidat(puntuacio)

            # guardar resultados dentro del diccionario del/a candidato/a
            candidatura["puntuacio"] = puntuacio
            candidatura["classificacio"] = classificacio

            # añadir a la lista final
            llista_candidats.append(candidatura)

            # mostrar resultado individual
            mostrar_individual(candidatura, puntuacio, classificacio)

        else:
            print("Candidatura rebutjada. Torna-ho a intentar.\n")

    # PASO 6: informe final
    mostrar_informe(llista_candidats)
```

```python
# COMPROBACIÓN DEL PASO 7

main_seleccio()
```

```python
# PASO 1 OBTENER FACTOR EDAD

# El objetivo de la función es convertir el grupo de edad en un factor numérico. Recibe un grupo de edad.
# La función hará la conversión a través de los diccionarios. Retornará un valor numérico.

def obtenir_factor_edat(grup_edat):
  return factors_edat[grup_edat]
```

```python
# COMPROBACIONES DEL PASO 1

print(f"El factor_edat per la franja '18-25' és: {obtenir_factor_edat("18-25")}")
print(f"El factor_edat per la franja '26-40' és: {obtenir_factor_edat("26-40")}")
print(f"El factor_edat per la franja '41-60' és: {obtenir_factor_edat("41-60")}")
print(f"El factor_edat per la franja '60+' és: {obtenir_factor_edat("60+")}")
```

```python
# PASO 2 OBTENER FACTOR Y CÓDIGO

# El objetivo de la función es convertir el código de comportamiento numérico en un segmento descriptivo y un factor de riesgo numérico. Recibe el código de comportamiento.
# La función hará la conversión a través de los diccionarios. Retornará un valor numérico y su descripción.

def obtenir_segment_factor(codi_comportament):
    return {
        0: {"segment": "Navega però no afegeix productes", "factor": 0.10},
        1: {"segment": "Afegeix productes però no checkout", "factor": 0.25},
        2: {"segment": "Arriba al checkout però no paga", "factor": 0.50},
        3: {"segment": "Cistella abandonada <24h", "factor": 0.75},
        4: {"segment": "Cistella abandonada >24h", "factor": 0.90}
    }[codi_comportament]
```

```python
# COMPROBACIONES DEL PASO 2

print(f"El codi de comportament per el valor '0' és: {obtenir_segment_factor(0)}")
print(f"El codi de comportament per el valor '1' és: {obtenir_segment_factor(1)}")
print(f"El codi de comportament per el valor '2' és: {obtenir_segment_factor(2)}")
print(f"El codi de comportament per el valor '3' és: {obtenir_segment_factor(3)}")
print(f"El codi de comportament per el valor '4' és: {obtenir_segment_factor(4)}")
```

```python
# PASO 3: CALCULAR EL RIESGO

# El objetivo de la función es calcular el riesgo final de cada cliente/a. Recibe dos datos numéricos: factor_codi y factor_edat
# La función realizará el cálculo siguiente: multiplicar el factor_codi por el factor_edat. Retornará un valor numérico.

def calcular_risc(factor_codi, factor_edat):
    return factor_codi * factor_edat
```

```python
# COMPROBACIONES DEL PASO 3

print(f"El risc d'aquest/a client/a és: {calcular_risc(0.90, 1.20)}")
print(f"El risc d'aquest/a client/a és: {calcular_risc(0.50, 1.00)}")
```

```python
# PASO 4: PROCESAR CLIENTES

# El objetivo de la función es combinar la información recabada hasta ahora. Recibe la lista de clientes
# La función realizará varias acciones para cada cliente/a (con bucle for).
# - Leerá el grupo de edad para definir el factor
# - Leerá el código de comportamiento para transformarlo en un valor numérico y su descripción
# - Calculará el riesgo en base a los códigos
# Retornará un diccionario de diccionarios (que se crea al principio)

def processar_clients(clients):

    clients_processats = {}

    for client in clients:

        id_client = client["id_client"]

        factor_edat = obtenir_factor_edat(client["grup_edat"])

        info_comportament = obtenir_segment_factor(client["codi_comportament"])
        segment = info_comportament["segment"]
        factors_codi = info_comportament["factor"]

        risc_final = calcular_risc(factors_codi, factor_edat)

        clients_processats[id_client] = {
            "nom": client["nom"],
            "edat": client["edat"],
            "grup_edat": client["grup_edat"],
            "codi_comportament": client["codi_comportament"],
            "segment": segment,
            "factors_codi": factors_codi,
            "factor_edat": factor_edat,
            "risc_final": risc_final
        }

    return clients_processats
```

pacients = [

nivells_risc = {

    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,

🎯 Objectius
L'objectiu és transformar aquestes dades en informació útil per als metges. A partir dels registres en brut, has de:

» Convertir cada string en una estructura neta (diccionari o similar).

```python
# PASO 0: INFORMACIÓN Y CREACIÓN DE DICCIONARIOS

pacients = [ "Maria|45|tos, Fiebre Alta, dolor_pecho", "Luis|33|dolor_cabeza,faTiga", "Sara|67|fiebre alta, dificultad RESPIRAR, tos", "Jordi|52|fatiga , tos", "Anna|29|fiebre alta,dolor_cabeza" ]

nivells_risc = { "fiebre_alta": 3, "dificultad_respirar": 5, "dolor_pecho": 4, "tos": 1, "fatiga": 1, "dolor_cabeza": 1 }
```

```python
# PASO 1: Estructurar la información

# El objetivo de la función es convertir cada string en una estructura limpia, concretamente un diccionario. Recibe un string
# La función separará la info donde encuentre el caracter "|" para obtener tres campos: nombre, edad y síntomas sin procesar
# Retornará un diccionario eliminando las separaciones por "|"

def separar_datos_paciente(registro):
    nom, edat, simptomes = registro.split("|")

    return {
        "nom": nom,
        "edat": int(edat),
        "simptomes": simptomes
    }
```

```python
# COMPROBACIÓN DEL PASO 1

for registro in pacients:
    print(separar_datos_paciente(registro))
```

```python
# PASO 2: Normalizar síntomas

# El objetivo de la función es normalizar los datos de síntomas recibidos, convirtiendo todas las letras a minúsculas, eliminando espacios y _ innecesarios y separándolos correctamente por comas
# La función usará lower, strip, replace y split para realizar esas acciones, respectivamente, normalizará el formato, y añadirá la información en una nueva lista "sintomas"
# Retornará el diccionario con el campo "sintomas" actualizado

def normalizar_sintomas(paciente):
    sintomas = paciente["simptomes"].lower().split(",")
    sintomas = [s.strip().replace(" ", "_") for s in sintomas]

    paciente["simptomes"] = sintomas

    return paciente
```

```python
# COMPROBACIÓN DEL PASO 2

paciente = separar_datos_paciente(pacients[0])
print(paciente)

paciente = normalizar_sintomas(paciente)
print(paciente)
```

```python
# PASO 3: Cálculo de riesgo

# El objetivo de la función es calcular/asignar un nivel de riesgo a cada paciente a partir de sus síntomas y del diccionario que nos ha sido dado de nivells_risc
# La función usará un bucle para recorrer la lista de síntomas del/a paciente y obtener su valor de riesgo (mediante get(), y si un síntoma no aparece, se asignará 0). Acumulará el riesgo en el contador que hemos creado.
# Retornará el diccionario con la clave risc_total actualizada

def calcular_riesgo(paciente):
    riesgo_total = 0

    for sintoma in paciente["simptomes"]:
        riesgo_total += nivells_risc.get(sintoma, 0)

    paciente["risc_total"] = riesgo_total

    return paciente
```

```python
# COMPROBACIÓN DEL PASO 3

paciente = separar_datos_paciente(pacients[0])
paciente = normalizar_sintomas(paciente)
paciente = calcular_riesgo(paciente)

print(paciente)
```

```python
# PASO 4: Orden por riesgo

# El objetivo de la función es ordenar los pacientes de mayor a menor riesgo. Recibe el listado de pacientes
# La función utiliza sorted en vez de sort para no modificar el listado original
# Retornará un listado ordenado

def ordenar_por_riesgo(lista_pacientes):
    return sorted(lista_pacientes,
                  key=lambda paciente: paciente["risc_total"],
                  reverse=True)
```

```python
# COMPROBACIÓN DEL PASO 4

from pprint import pprint

pacientes_procesados = []

for registro in pacients:
    paciente = separar_datos_paciente(registro)
    paciente = normalizar_sintomas(paciente)
    paciente = calcular_riesgo(paciente)

    pacientes_procesados.append(paciente)

pprint(pacientes_procesados)

pacientes_ordenados = ordenar_por_riesgo(pacientes_procesados)

pprint(pacientes_ordenados)
```

```python
# PASO 5: Main, orquestradora

# El objetivo de la función es coordinar las otras funciones que tratan los datos de los pacientes.
# En primer lugar crea una lista vacía para almacenar los pacientes ya procesados. Recorre la lista original y para cada paciente:
# Separa los datos del registro y los convierte en un diccionario, normaliza los síntomas para unificar el formato, calcula el riesgo total
# Retornará un listado ordenado segun el nivel de riesgo

def main():
    pacientes_procesados = []

    for registro in pacients:
        paciente = separar_datos_paciente(registro)
        paciente = normalizar_sintomas(paciente)
        paciente = calcular_riesgo(paciente)

        pacientes_procesados.append(paciente)

    pacientes_ordenados = ordenar_por_riesgo(pacientes_procesados)

    print("PACIENTES ORDENADOS POR PRIORIDAD\n")

    for paciente in pacientes_ordenados:
        print(paciente)
```

```python
# PASO 0: INFORMACIÓN Y CREACIÓN DE DICCIONARIOS

clients = [ {"nom": "Ana", "edat": 42, "codi": 0}, {"nom": "Carlos", "edat": 30, "codi": 2}, {"nom": "Isabel", "edat": 55, "codi": 3}, {"nom": "Jorge", "edat": 40, "codi": 1}, {"nom": "Marta", "edat": 28, "codi": 0} ]

nivells_seguro = { 0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01}, 1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05}, 2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15}, 3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30} }
```

```python
# PASO 1: Clasificar clientes

# El objetivo de la función es devolver información sobre los clientes en base a su código de comportamiento, que encontramos en el diccionario nivells_seguro. Recibe un cliente de la lista y el diccionario dado.
# La función crea nuevas variables en los clientes, concretamente: categoria, preu_base, fraude
# La función retorna el diccionario actualizado

def afegir_cat_pb_f(client, nivells_seguro):
    nivell_seguro = nivells_seguro[client["codi"]]

    client["categoria"] = nivell_seguro["categoria"]
    client["preu_base"] = nivell_seguro["preu_base"]
    client["fraude"] = nivell_seguro["fraude"]

    return client
```

```python
# COMPROBACIÓN DEL PASO 1

cliente_prueba = {"nom": "Natalia", "edat": 30, "codi": 0}

resultado = afegir_cat_pb_f(cliente_prueba, nivells_seguro)

print(resultado)
```

```python
# PASO 2: Calcular el precio final

# El objetivo de la función es calcular el precio final. Recibe el precio base y la edad
# La función realizará la siguiente operación: preu_final = preu_base * (1 + edat/100)
# Retornará el resultado del cálculo

def calcular_preu_final(preu_base, edat):

  preu_final = round(preu_base * (1 + edat/100), 2)

  return preu_final
```

```python
# COMPROBACIÓN DEL PASO 2

calcular_preu_final(120, 42)
```

```python
# PASO 3, Añadir precio final

# El objetivo de la función es añadir el precio final a la información del cliente. Recibe un cliente
# La función añade una nuvea clave al diccionario
# Retornará el cliente enriquecido

def afegir_preu_final(client):
    client["preu_final"] = calcular_preu_final(client["preu_base"], client["edat"])

    return client
```

```python
# COMPROBACIÓN DEL PASO 3

cliente_prueba = {
    "nom": "Carlos",
    "edat": 30,
    "codi": 2,
    "categoria": "Risc alt",
    "preu_base": 350,
    "fraude": 0.15
}

resultado = afegir_preu_final(cliente_prueba)

print(resultado)
```

```python
# PASO 4: Crear diccionario con la información actualizada

# El objetivo de la función es crear un diccionario con todos los datos de todos los clientes actualizados. Recibe una lista
# Mediante un bucle for, recorre cada cliente para añadirlo todo en una nueva lista
# Retorna un diccionario actualizado

def processar_clients():
    clients_processats = {}

    for client in clients:
        afegir_cat_pb_f(client, nivells_seguro)
        afegir_preu_final(client)

        clients_processats[client["nom"]] = client

    return clients_processats
```

```python
# PASO 5, ordenar clientes

# El objetivo de la función es ordenar los clientes en base a la probabilidad de fraude y el precio. Recibe el diccionario de clientes ya procesados
# La función utilizará sorted para la ordenación
# Devolverá una lista ordenada

def ordenar_clients(clients_processats):

    ordenats = sorted(
        clients_processats.items(),
        key=lambda x: (x[1]["fraude"], x[1]["preu_final"]),
        reverse=True
    )
    return ordenats
```

```python
# PASO 1: Obtenir paràmetres permesos

# En esta función se determinan los valores permitidos. Recibe los parámetros que queremos incluir.
# La función retornará una cadena con los valores selecciondos

def obtenir_caracters_permesos(majuscules, minuscules, numeros, signes):
    caracters = ""

    if majuscules:
        caracters += "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    if minuscules:
        caracters += "abcdefghijklmnopqrstuvwxyz"

    if numeros:
        caracters += "0123456789"

    if signes:
        caracters += "!@#$"

    return caracters
```

```python
# PASO 2: Afegir caràcters obligatoris

# El objetivo de la función es hacer que cumpla los criterios. Recibe los criterios que tendrá la contraseña
# La función creará una lista donde añadirá un caracter de cada grupo necesario
# Retornará una lista con los valores

import numpy as np

def añadir_caracteres_obligatorios(majuscules, minuscules, numeros, signes):

  valors = []

  if majuscules is True:
    valors.append(str(np.random.choice(list("ABCDEFGHIJKLMNOPQRSTUVWXYZ"))))

  if minuscules is True:
    valors.append(str(np.random.choice(list("abcdefghijklmnopqrstuvwxyz"))))

  if numeros is True:
    valors.append(str(np.random.choice(list("0123456789"))))

  if signes is True:
    valors.append(str(np.random.choice(list("!@#$"))))

  return valors
```

```python
# PASO 3: Omplir contrassenya

# El objetivo de esta función es "rellenar" la contrasenya para llegar al nº de valores solicitados (longitud)
# Recibe la lista de carácteres obligatorios (valors), comprueba su longitud. A través de bucle while, mientras no se llegue a la longitud deseada, se escoge un carácter y se añade a la lista
# Devuelve la lista completa
# Opto por no modificar la lista original

def omplir_contrassenya(valors, caracters, longitud):
    completa = valors.copy()

    while len(completa) < longitud:
        caracter = str(np.random.choice(list(caracters)))
        completa.append(caracter)

    return completa
```

```python
# COMPROBACIÓN DEL PASO 3

caracters = obtenir_caracters_permesos(
    majuscules=True,
    minuscules=True,
    numeros=True,
    signes=False
)

valors = añadir_caracteres_obligatorios(
    majuscules=True,
    minuscules=True,
    numeros=True,
    signes=False
)

resultado = omplir_contrassenya(valors, caracters, 10)

print(resultado)
```

```python
# PASO 4: Barrejar valors

# El objetivo de la función es mezclar aleatoriamente los valores de la lista
# Recibe la lista, mediante shuffle se mezcla, devolverá una lista
# Opto por no modificar la lista original

import numpy as np

def barrejar_valors(completa):

    barrejada = completa.copy()

    np.random.shuffle(barrejada)

    return barrejada
```

```python
# PASO 5: Validación de la contraseña

# El objetivo de la función es validar que se cumplen los criterios de longitud y de un símbolo de cada. Recibe la lista (llista barrejada) y la longitud
# La función recorrerá cada carácter para comprobar si se cumple o no con un bucle for
# Si es correcto, retornará un mensaje de validación, si es incorrecto, uno de error

def validar(barrejada, longitud):
    if len(barrejada) < longitud:
        print(f"La contrassenya ha de tenir almenys {longitud} caràcters")
        return False

    tiene_mayuscula = False
    tiene_minuscula = False
    tiene_numero = False
    tiene_simbolo = False

    for caracter in barrejada:
        if caracter.isupper():
            tiene_mayuscula = True
        elif caracter.islower():
            tiene_minuscula = True
        elif caracter.isdigit():
            tiene_numero = True
        else:
            tiene_simbolo = True

    if not tiene_mayuscula:
        print("Falta una mayúscula")
    if not tiene_minuscula:
        print("Falta una minúscula")
    if not tiene_numero:
        print("Falta un número")
    if not tiene_simbolo:
        print("Falta un símbolo")

    if (tiene_mayuscula and tiene_minuscula and
            tiene_numero and tiene_simbolo):
        print("La contraseña es válida")

    return (
        tiene_mayuscula and
        tiene_minuscula and
        tiene_numero and
        tiene_simbolo
    )
```

```python
# PASO 6: Main, orquestradora, generar contrassenya

# El objetivo de la función es coordinar todos los pasos anteriores
# La función define los carácteres permitidos, añade un carácter obligatorio de cada tipo, rellena la contraseña hasta la longitud deseada, mezcla esa lista y valida que se cumplan los criterios
# La función retorna un listado y su validación

import numpy as np

def generar_contrassenya():
    longitud = 10
    majuscules = True
    minuscules = True
    numeros = True
    signes = True

    caracters = obtenir_caracters_permesos(
        majuscules, minuscules, numeros, signes
    )

    valors = añadir_caracteres_obligatorios(
        majuscules, minuscules, numeros, signes
    )

    completa = omplir_contrassenya(
        valors, caracters, longitud
    )

    barrejada = barrejar_valors(completa)

    contrassenya = "".join(barrejada)

    print("Contrasenya generada:", contrassenya)

    validar(contrassenya, longitud)
```

**» 4.La classificació final ordenada per punts on cada victòria suma 3 pts, cada empat 1 pts y cada derrota 0 pts.**

```python
# PASO 0: Importar y leer archivo

# Leemos el archivo línea por línea mediante bucle for
# Muestra el contenido en pantalla

def lectura():
  with open("historic partits.txt", "r", encoding="utf-8") as archivo:
    lineas = []
    for linea in archivo:
        lineas.append(linea)
    return lineas
```

```python
# COMPROBACIÓN

from pprint import pprint

pprint(lineas[:3])
```

```python
# PASO 1: Limpiar formato

 # El objetivo de la función es "limpiar" el formato, consiguiendo que cada línea sea un string sin espacios ni tabulaciones ("\t")

def preparar_partidos(lineas):
    partidos = []

    for linea in lineas:
        linea = linea.strip()

        datos = linea.split("\t")

        local = datos[0]
        resultado = datos[1]
        visitante = datos[2]

        goles_local, goles_visitante = resultado.split("-")

        partidos.append([
            local,
            int(goles_local),
            int(goles_visitante),
            visitante
        ])

    return partidos
```

```python
# COMPROBACIÓN PASO 1

preparar_partidos(lineas)
```

```python
# PASO 2: Crear diccionario

# El objetivo de la función es crear los distintos equipos donde más tarde se haran los recuentos. Recibe la lista de partidos
# Mediante bucle for, pediremos la posición 1 ([0]), referente al equipo local y la 4 ([3]) referente al equipo visitante para consultar si los equipos que van apareciendo ya "tienen una ficha creada" en nuestro diccionario o todavía no
# Devolverá el diccionario actualizado

def crear_equipos(partidos):
    equipos = {}

    for partido in partidos:
        local = partido[0]
        visitante = partido[3]

        if local not in equipos:
            equipos[local] = {
                "GF": 0,
                "GC": 0,
                "Puntos": 0
            }

        if visitante not in equipos:
            equipos[visitante] = {
                "GF": 0,
                "GC": 0,
                "Puntos": 0
            }

    return equipos
```

```python
# COMPROBACIÓN DEL PASO 2

partidos = preparar_partidos(lineas)

crear_equipos(partidos)
```

```python
# PASO 3: Procesar goles

# El objetivo de la función es sacar la información de las líneas para añadirlas al diccionario correspondiente, primero centrándonos solo en los goles. Recbie el listado de partidos
# Mediante un bucle for, recorrerá cada linea para añadir los goles que ha marcado cada equipo y completar así ambos campos (goles a favor y goles en contra).
# La función retornará el diccionario actualizado

def procesar_goles(partidos):

  equipos = crear_equipos(partidos)

  for partido in partidos:
    local = partido[0]
    visitante = partido[3]

    goles_local = partido[1]
    goles_visitante = partido[2]

    equipos[local]["GF"] += goles_local
    equipos[local]["GC"] += goles_visitante
    equipos[visitante]["GF"] += goles_visitante
    equipos[visitante]["GC"] += goles_local

  return equipos
```

```python
# COMPROBACIÓN DEL PASO 3

from pprint import pprint

equipos = procesar_goles(partidos)

pprint(equipos)
```

```python
# PASO 4: Procesar la puntuación

# El objetivo de la función es sacar la información de las líneas para añadirlas al diccionario correspondiente, ahora centrándonos en los puntos. Recbie el listado de partidos y equipos
# Mediante un bucle for, recorrerá cada linea para contabilizar los goles y actualizar el campo de puntos
# La función retornará el diccionario actualizado

def procesar_puntos(partidos, equipos):

    for partido in partidos:

        local = partido[0]
        goles_local = partido[1]
        goles_visitante = partido[2]
        visitante = partido[3]

        if goles_local > goles_visitante:
            equipos[local]["Puntos"] += 3

        elif goles_local < goles_visitante:
            equipos[visitante]["Puntos"] += 3

        else:
            equipos[local]["Puntos"] += 1
            equipos[visitante]["Puntos"] += 1

    return equipos
```

```python
# COMPROBACIÓN DEL PASO 4

from pprint import pprint

equipos = crear_equipos(partidos)

equipos = procesar_goles(partidos)

equipos = procesar_puntos(partidos, equipos)

pprint(equipos)
```

```python
# PASO 5: Buscar al equipo más goleador

# El objetivo de la función es buscar el equipo que más goles haya anotado. Recibe el diccionario de equipos
# Se crea un contador a 0 y, a través de un bucle for, vamos recorriendo cada equipo para quedarnos con el número más grande
# Retornará el equipo y el contador

def equipo_mas_goleador(equipos):
    max_goles_f = 0
    equipo = ""

    for nombre, datos in equipos.items():
        if datos["GF"] > max_goles_f:
            max_goles_f = datos["GF"]
            equipo = nombre

    return equipo, max_goles_f
```

```python
# COMPROBACIÓN DEL PASO 5

mas_goleador = equipo_mas_goleador(equipos)

print(f"El equipo más goleador es el {mas_goleador[0]}, con un total de {mas_goleador[1]} goles")
```

```python
# PASO 6: Buscar al equipo más goleado

# El objetivo de la función es buscar el equipo que más goles haya recibido. Recibe el diccionario de equipos
# Se crea un contador a 0 y, a través de un bucle for, vamos recorriendo cada equipo para quedarnos con el número más grande
# Retornará el equipo y el contador

def equipo_mas_goleado(equipos):
    max_goles_c = 0
    equipo = ""

    for nombre, datos in equipos.items():
        if datos["GC"] > max_goles_c:
            max_goles_c = datos["GC"]
            equipo = nombre

    return equipo, max_goles_c
```

```python
# COMPROBACIÓN DEL PASO 6

mas_goleado = equipo_mas_goleado(equipos)

print(f"El equipo más goleado es el {mas_goleado[0]}, con un total de {mas_goleado[1]} goles")
```

```python
# PASO 7: Contador de goles por equipo

# El objetivo de la función es contar los goles a favor de cada equipo

def ordenar_por_puntos(equipos):
    return sorted(equipos.items(), key=lambda x: x[1]["Puntos"], reverse=True)
```

```python
# COMPROBACIÓN PASO 7

ranking_goles = ordenar_por_puntos(equipos)

for equipo, datos in ranking_goles:
    print(equipo, datos["Puntos"])
```

```python
# PASO 8: Ordenar los resultados

# El objetivo de la función es ordenar los equipos que recibe (usando sorted para no modificar el original) en base a los puntos que gana
# Retornará

def ordenar_clasificacion(equipos):
    clasificacion = sorted(
        equipos.items(),
        key=lambda x: x[1]["Puntos"],
        reverse=True
    )

    return clasificacion
```

```python
# COMPROBACIÓN DEL PASO 8

ordenar_clasificacion(equipos)
```

```python
# PASO 9, Main, informe final

# El objetivo de la función es mostrar los resultados de la clasificación, ordenados y legibles, pudiendo responder a las preguntas que se nos han hecho.

def main_partidos():

    lineas_partidos = lectura()

    partidos = preparar_partidos(lineas_partidos)

    equipos = crear_equipos(partidos)

    equipos = procesar_goles(partidos)
    equipos = procesar_puntos(partidos, equipos)

    mas_goleador = equipo_mas_goleador(equipos)
    mas_goleado = equipo_mas_goleado(equipos)

    ranking_goles = ordenar_por_puntos(equipos)

    clasificacion = ordenar_clasificacion(equipos)

    print("\nEQUIPO MÁS GOLEADOR:")
    print("--------------------")
    print(mas_goleador[0], "->", mas_goleador[1], "goles")

    print("\nEQUIPO MÁS GOLEADO:")
    print("--------------------")
    print(mas_goleado[0], "->", mas_goleado[1], "goles recibidos")

    print("\nGOLES POR EQUIPO:")
    print("--------------------")
    for equipo, datos in ranking_goles:
     print(equipo, datos["GF"])

    print("\nCLASIFICACIÓN FINAL:")
    print("--------------------")

    for posicion, equipo in enumerate(clasificacion, start=1):
        print(
            posicion,
            equipo[0],
            "-",
            equipo[1]["Puntos"],
            "puntos"
        )
```
